```python
# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()# -*- coding: utf-8 -*-
import arcpy
import socket
import re

def get_ip_address():
    """获取本机IP地址"""
    try:
        # 创建一个UDP套接字
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        # 连接到一个远程地址（不会真正发送数据）
        s.connect(("8.8.8.8", 80))
        # 获取本地IP地址
        ip_address = s.getsockname()[0]
        s.close()
        return ip_address
    except Exception as e:
        print(f"获取IP地址时出错: {e}")
        return None

def find_layer_by_ip(mxd, ip_address):
    """在地图中查找包含IP地址的图层"""
    # 将IP地址转换为搜索模式（例如192.168.1.1 -> 192.168.1.%）
    ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
    
    # 获取当前地图文档
    layers = arcpy.mapping.ListLayers(mxd)
    
    for layer in layers:
        if layer.supports("DATASOURCE"):
            # 检查图层名称或数据源是否包含IP相关信息
            if "ip" in layer.name.lower() or "address" in layer.name.lower():
                return layer
            
            # 检查图层字段是否包含IP地址字段
            fields = arcpy.ListFields(layer)
            for field in fields:
                if "ip" in field.name.lower() or "address" in field.name.lower():
                    return layer
    
    return None

def zoom_to_ip_location(mxd, df, ip_layer, ip_address):
    """缩放到IP地址对应的位置"""
    try:
        # 创建查询表达式
        ip_pattern = ".".join(ip_address.split(".")[:3]) + ".%"
        where_clause = f"IP_ADDRESS LIKE '{ip_pattern}'"  # 根据实际字段名调整
        
        # 选择匹配IP地址的特征
        arcpy.SelectLayerByAttribute_management(ip_layer, "NEW_SELECTION", where_clause)
        
        # 获取选择集的范围
        desc = arcpy.Describe(ip_layer)
        if desc.FIDSet:  # 如果有选中的要素
            # 缩放到选择集
            df.extent = ip_layer.getSelectedExtent()
            # 放大一点以便更好地查看
            df.scale *= 0.9
            print(f"已缩放到IP地址 {ip_address} 的位置")
        else:
            print(f"未找到与IP地址 {ip_address} 匹配的位置")
            
    except Exception as e:
        print(f"缩放到IP位置时出错: {e}")

def main():
    """主函数"""
    try:
        # 获取当前地图文档
        mxd = arcpy.mapping.MapDocument("CURRENT")
        df = arcpy.mapping.ListDataFrames(mxd)[0]  # 获取第一个数据框
        
        # 获取IP地址
        ip_address = get_ip_address()
        if not ip_address:
            print("无法获取IP地址")
            return
        
        print(f"本机IP地址: {ip_address}")
        
        # 查找包含IP地址的图层
        ip_layer = find_layer_by_ip(mxd, ip_address)
        if not ip_layer:
            print("未找到包含IP地址的图层")
            return
        
        print(f"找到IP图层: {ip_layer.name}")
        
        # 缩放到IP地址位置
        zoom_to_ip_location(mxd, df, ip_layer, ip_address)
        
        # 刷新地图
        arcpy.RefreshActiveView()
        arcpy.RefreshTOC()
        
    except Exception as e:
        print(f"处理过程中出错: {e}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()
```
